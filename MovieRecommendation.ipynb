```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv', engine='python')
```

```python
movies.sample(1)
```

```python
credits.sample(1)
```

```python
movies = movies.merge(credits, on='title')
```

```python
movies.info()
```

```python
movies = movies[['genres','cast','crew','keywords','overview','title','movie_id','vote_average','popularity']]
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
def convert(obj):
  l = []
  for(i) in eval(obj):
    l.append(i['name'])
  return l
```

```python
convert(movies.iloc[0].genres)
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies.head()
```

```python
movies['cast'].iloc[1]
```

```python
def convert3(obj):
  l = []
  counter = 0
  for(i) in eval(obj):
    if counter != 3:
      l.append(i['name'])
      counter += 1
    else:
      break
  return l
```

```python
movies['cast'] = movies['cast'].apply(convert3)
```

```python
movies['crew'].iloc[0]
```

```python
def convertCrew(obj):
  l = []
  for(i) in eval(obj):
    if i['job'] == 'Director':
      l.append(i['name'])
      break
  return l
```

```python
movies['crew'] = movies['crew'].apply(convertCrew)
```

```python
movies.head()
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

```python
movies.isnull().sum()
```

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies.head()
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
movies_new = movies[['movie_id','title','tags','vote_average','popularity']]
```

```python
movies_new.head()
```

```python
movies_new['tags'] = movies_new['tags'].apply(lambda x:" ".join(x))
```

```python
movies_new['tags'] = movies_new['tags'].apply(lambda x:x.lower())
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
  y = []
  for i in text.split():
    y.append(ps.stem(i))
  return " ".join(y)
```

```python
movies_new['tags'] = movies_new['tags'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
cv.get_feature_names_out()
```

```python
vectors = cv.fit_transform(movies_new['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

```python
def recommend(movie):
  movie_index = movies_new[movies_new['title'] == movie].index[0]
  distances = similarity[movie_index]
  movies_list = sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]
  for i in movies_list:
    movie = movies.iloc[i[0]]
    print(movie['title'], movie['vote_average'], movie['popularity'])
```

```python
recommend('The Dark Knight Rises')
```
